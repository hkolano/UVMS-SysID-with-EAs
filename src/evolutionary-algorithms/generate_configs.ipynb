```python
import yaml
import os
import copy
```

```python
yaml_dir = "/home/gonzaeve/UVMS-SysID-with-EAs/src/evolutionary-algorithms/config/default.yaml"
with open(yaml_dir) as stream:
    default = yaml.safe_load(stream)
```

```python
new_dir = "/home/gonzaeve/UVMS-SysID-with-EAs/src/evolutionary-algorithms/config/alpha"
if not os.path.exists(new_dir):
    os.makedirs(new_dir)
```

```python
n = list(range(20))
for ind in n:
    # Make the yaml file
    new = copy.deepcopy(default)
    save_dir = "~/hpc-share/uvms/preliminary/alpha/trial_"+str(ind)
    new['save_options']['save_dir'] = save_dir
    new['ea_parameters']['start_population']['option']='default_params'
    new['ea_parameters']['start_population']['load_dir'] = ''
    # Save it
    config_dir = new_dir+"/trial_"+str(ind)+".yaml"
    with open(config_dir, 'w') as outfile:
        yaml.dump(new, outfile, default_flow_style=False)
```

```python
"""Generate the configs for beta"""
yaml_dir = "/home/gonzaeve/UVMS-SysID-with-EAs/src/evolutionary-algorithms/config/default.yaml"
with open(yaml_dir) as stream:
    default = yaml.safe_load(stream)
new_dir = "/home/gonzaeve/UVMS-SysID-with-EAs/src/evolutionary-algorithms/config/beta"
if not os.path.exists(new_dir):
    os.makedirs(new_dir)

n = list(range(20))
for ind in n:
    # Make the yaml file
    new = copy.deepcopy(default)
    save_dir = "~/hpc-share/uvms/preliminary/beta/trial_"+str(ind)
    new['save_options']['save_dir'] = save_dir
    new['ea_parameters']['start_population']['option']='random'
    new['ea_parameters']['start_population']['load_dir'] = ''
    # Save it
    config_dir = new_dir+"/trial_"+str(ind)+".yaml"
    with open(config_dir, 'w') as outfile:
        yaml.dump(new, outfile, default_flow_style=False)
```
